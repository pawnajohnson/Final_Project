#### Checking the churn rate based on gender 

```python
df['Churn_Numeric'] = df['Churn'].map({'Yes': 1, 'No': 0})
df.groupby('gender')['Churn_Numeric'].mean() * 100
```

#### To understand if Internet Service affects the churn rate

```python
multi_group = df.groupby(['Contract', 'InternetService'])['Churn_Numeric'].mean() * 100
print(multi_group.round(2))
```

#### Compare the financial profile of customers who leave versus those who stay by running numerical aggregations (mean and median).

```python
df.groupby('Churn')[['MonthlyCharges', 'TotalCharges']].agg(['mean', 'median', 'min', 'max']).round(2)
```

#### To understand if the payment method affects the churn rate

```python
payment_risk = df.groupby('PaymentMethod')['Churn_Numeric'].mean() * 100
print(payment_risk.sort_values(ascending=False).round(2))
```

#### Output - The output suggestes that the highest churn rate is among customers who pay their bills through electronic check because this involves the customer to reach a "decision point" on a monthly basis where they evaluate a bill and consider cancelling . While the automatic check system have a recurring billling system which automatically transfers the monthly bill from the customers account so the customer at that point does not have to make a decision. Mailed check users are somewhere close to the electronic check customer churn rate yet are lower than them this suggests that electronic check users are usually price sensitive subscribers.

#### Grouping customers based on bins to check at what point exactly a customer is churned

```python
bins = [-1, 12, 36, 72]
labels = ['0-1 Year (New)', '1-3 Years (Mid)', '3+ Years (Loyal)']
df['Tenure_Bracket'] = pd.cut(df['tenure'], bins=bins, labels=labels)


tenure_analysis = df.groupby('Tenure_Bracket')['Churn_Numeric'].agg(['count', 'mean'])
tenure_analysis['Churn_Rate_%'] = (tenure_analysis['mean'] * 100).round(2)
print(tenure_analysis[['count', 'Churn_Rate_%']])
```

#### Output - This output shows that customers churn is the highest amongst the new customers while the customers between 1-3 years tenure have lesser churn and , customers with 3+ years tenure have the lowest churn.

### 1. PROJECT OVERVIEW AND OBJECTIVES
#### - Analyzed a telecommunications dataset of 7043 customer records across 21 atttributes (demographics, services,contract terms,tenure and billing)
#### - Goal :  To identify the primary drivers of customer churn and translate those insights into actionable retention strategies.

### 2. DATA CLEANING & PREPROCESSING
#### - Data Integrity: Verified 0 duplicate records and confirmed logical consistency  across service columns (all 682 customers without phone service were properly labeled as 'No phone service').
#### - Text Standardization: Cleaned and standardized formatting (title case & whitespace removal) for customerID, gender, Contract, and PaymentMethod.
#### - Missing Value Imputation: Converted TotalCharges to numeric format and imputed 11 missing values (all belonging to new customers with tenure = 0) using their respective MonthlyCharges.
#### - Feature Engineering: Created a binary Churn_Numeric target column and segmented tenure into three brackets: '0-1 Year', '1-3 Years', and '3+ Years'.

### 3. KEY ANALYTICAL FINDINGS
#### - Overall Churn Rate: 26.54% of customers (1,869) churned, while 73.46% (5,174) were retained.
#### - Contract & Internet Service: Month-to-Month Fiber Optic customers have the highest churn rate at 54.61%, compared to 32.22% for Month-to-Month DSL and only 0.78% for Two-Year contract customers without internet service.
#### - Tenure Impact: Churn is heavily front-loaded—47.44% for new customers (0-1 Year), dropping to 25.54% (1-3 Years), and 11.93% (3+ Years).
#### - Billing Sensitivity: Churned customers pay higher average monthly bills ($74.44 mean / $79.65 median) than retained customers ($61.27 mean / $64.43  median), though their cumulative TotalCharges are lower ($1,531.80 vs $2,550.00).
#### - Payment Method Friction: Electronic Check users churn at 45.29%—more than double Mailed Check (19.11%) and nearly triple automated methods like Bank  Transfer (16.71%) and Credit Card (15.24%).
#### - Gender Neutrality: Gender shows negligible impact on churn (26.92% Female vs. 26.16% Male).

### 4. STRATEGIC RECOMMENDATIONS
#### - Incentivize Contract Upgrades: Offer discounts or bundled add-ons (Tech Support, Online Security) to convert Month-to-Month Fiber Optic users into  1- or 2-year contracts.
#### - Strengthen First-Year Onboarding: Introduce proactive check-ins and milestone  rewards during the first 12 months to reduce early-stage attrition.
#### - Promote Auto-Pay Enrollment: Provide small bill credits to encourage Electronic Check users to switch to automatic credit card or bank transfers.
#### - Audit Fiber Optic Value: Review Fiber Optic pricing and service reliability to reduce dissatisfaction among high-paying subscribers.
